50. Write a Python program that uses multiple built-in functions together to process a list of numbers.

```python
a = [10, 20, 30, 40, 50]
list_sum = sum(a)
list_length = len(a)
list_max = max(a)
list_min = min(a)
print(f"Original list: {a}")
print(f"Sum of the list: {list_sum}")
print(f"Length of the list: {list_length}")
print(f"Maximum value in the list: {list_max}")
print(f"Minimum value in the list: {list_min}")
```

```python
# 30. Create a scientific calculator that supports square root, factorial,
# logarithm, power, trigonometric functions, GCD, and LCM.
import math

def scientific_calculator():
    while True:
        print("\n===== Scientific Calculator =====")
        print("1. Square Root")
        print("2. Factorial")
        print("3. Logarithm (Natural Log)")
        print("4. Power")
        print("5. Sine")
        print("6. Cosine")
        print("7. Tangent")
        print("8. GCD")
        print("9. LCM")
        print("10. Exit")

        try:
            choice = int(input("Enter your choice (1-10): "))

            if choice == 1:
                num = float(input("Enter a number: "))
                print(f"Square Root = {math.sqrt(num)}")
            elif choice == 2:
                num = int(input("Enter an integer: "))
                if num < 0:
                    print("Factorial is not defined for negative numbers.")
                else:
                    print(f"Factorial = {math.factorial(num)}")
            elif choice == 3:
                num = float(input("Enter a number: "))
                if num <= 0:
                    print("Logarithm is not defined for non-positive numbers.")
                else:
                    print(f"Natural Logarithm = {math.log(num)}")
            elif choice == 4:
                base = float(input("Enter the base number: "))
                exponent = float(input("Enter the exponent: "))
                print(f"Result = {math.pow(base, exponent)}")
            elif choice == 5:
                degrees = float(input("Enter angle in degrees: "))
                print(f"Sine = {math.sin(math.radians(degrees))}")
            elif choice == 6:
                degrees = float(input("Enter angle in degrees: "))
                print(f"Cosine = {math.cos(math.radians(degrees))}")
            elif choice == 7:
                degrees = float(input("Enter angle in degrees: "))
                print(f"Tangent = {math.tan(math.radians(degrees))}")
            elif choice == 8:
                num1 = int(input("Enter first number: "))
                num2 = int(input("Enter second number: "))
                print(f"GCD = {math.gcd(num1, num2)}")
            elif choice == 9:
                num1 = int(input("Enter first number: "))
                num2 = int(input("Enter second number: "))
                print(f"LCM = {math.lcm(num1, num2)}")
            elif choice == 10:
                print("Exiting calculator. Goodbye!")
                break
            else:
                print("Invalid choice. Please enter a number between 1 and 10.")
        except ValueError:
            print("Invalid input. Please enter a valid number.")
        except Exception as e:
            print(f"An error occurred: {e}")

# Run the calculator
scientific_calculator()
```

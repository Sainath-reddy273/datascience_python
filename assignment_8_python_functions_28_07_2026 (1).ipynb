## Assignment 35: Fibonacci Function

Write a function named `generate_fibonacci()`.

### Requirements

- Accept the number of terms.
- Return the complete Fibonacci series as a list.
- Handle zero and negative term counts.
- Do not use recursion in this assignment.

```python
def generate_fibonacci(n):
    a, b = 0, 1
    series = []
    for i in range(n):
        series.append(a)
        a, b = b, a + b
    return series
print(generate_fibonacci(10))
```

Return multiple functions:

    return deposit, withdraw, check_balance

```python
def create_bank_account(balance):
    def deposit(amount):
        nonlocal balance
        balance += amount
        print("Deposited:", amount)
    def withdraw(amount):
        nonlocal balance
        if amount <= balance:
            balance -= amount
            print("Withdrawn:", amount)
        else:
            print("Insufficient Balance")
    def check_balance():
        print("Balance:", balance)
    return deposit, withdraw, check_balance
deposit, withdraw, check_balance = create_bank_account(1000)
check_balance()
deposit(500)
check_balance()
withdraw(300)
check_balance()
withdraw(1500)
check_balance()
```

- Preserve the original function's return value.

### Expected Pattern

    Starting function...
    Result: 30
    Function completed.
    

```python
def execution_logger(func):
    def wrapper(a, b):
        print("Starting function...")
        result = func(a, b)
        print("Function completed.")
        return result
    return wrapper
@execution_logger
def add(a, b):
    return a + b
result = add(10, 20)
print("Result:", result)
```

```python
from functools import reduce

employees = [
    {"id":101,"name":"Ravi","department":"IT","salary":60000,"performance_score":88,"experience":5},
    {"id":102,"name":"Anitha","department":"HR","salary":45000,"performance_score":72,"experience":4},
    {"id":103,"name":"Kiran","department":"IT","salary":75000,"performance_score":94,"experience":7},
    {"id":104,"name":"Meena","department":"Finance","salary":55000,"performance_score":81,"experience":6},
    {"id":105,"name":"Arjun","department":"IT","salary":40000,"performance_score":65,"experience":2}
]

# ----------------- Function -----------------
def employee_report():

    # Nested Function
    def show_title():
        print("Employee Performance Report")

    show_title()

    # Filter()
    top_employees = list(filter(lambda emp: emp["performance_score"] >= 80, employees))

    print("\nTop Employees:")
    for emp in top_employees:
        print(emp["name"], emp["performance_score"])

    # Map()
    bonus = list(map(lambda emp:
                     (emp["name"], emp["salary"] * 0.10),
                     top_employees))

    print("\nBonus (10%):")
    for b in bonus:
        print(b)

    # Reduce()
    total_salary = reduce(
        lambda total, emp: total + emp["salary"],
        employees,
        0
    )

    print("\nTotal Salary:", total_salary)

    # Lambda Sorting
    sorted_emp = sorted(
        employees,
        key=lambda emp: emp["salary"],
        reverse=True
    )

    print("\nEmployees by Salary:")
    for emp in sorted_emp:
        print(emp["name"], emp["salary"])


# ----------------- Closure -----------------
def bonus_calculator(percent):

    def calculate(salary):
        return salary * percent / 100

    return calculate


bonus20 = bonus_calculator(20)

print("20% Bonus for 50000 =", bonus20(50000))

employee_report()
```

## Assignment 35: Fibonacci Function

Write a function named `generate_fibonacci()`.

### Requirements

- Accept the number of terms.
- Return the complete Fibonacci series as a list.
- Handle zero and negative term counts.
- Do not use recursion in this assignment.

```python
def generate_fibonacci(n):
    a, b = 0, 1
    series = []
    for i in range(n):
        series.append(a)
        a, b = b, a + b
    return series
print(generate_fibonacci(10))
```

## Assignment 56: Mask Email Addresses

Given a list of email addresses, mask the username portion.

### Example

    bharath@gmail.com → b******@gmail.com

### Requirements

- Use `map()`.
- Keep the first character of the username.
- Replace remaining username characters with `*`.
- Keep the domain unchanged.
- Create a separate named function for masking.

```python
def mask_email(email):
    username, domain = email.split("@")
    return username[0] + "*" * (len(username) - 1) + "@" + domain
emails = input("Enter email addresses: ").split()
masked_emails = list(map(mask_email, emails))
print("Original Emails:", emails)
print("Masked Emails:", masked_emails)
```

## Assignment 57: Format Phone Numbers

Given a list of 10-digit phone numbers as strings, format each one.

### Expected Format

    9876543210 → 987-654-3210

### Requirements

- Use `map()`.
- Validate the length and digits.
- Return `"Invalid"` for invalid phone numbers.

```python
def format_phone(phone):
    if len(phone) == 10 and phone.isdigit():
        return phone[:3] + "-" + phone[3:6] + "-" + phone[6:]
    else:
        return "Invalid"
phones = input("Enter phone numbers: ").split()
formatted_phones = list(map(format_phone, phones))
print("Original Numbers:", phones)
print("Formatted Numbers:", formatted_phones)
```
